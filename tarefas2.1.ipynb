### Exercicio 1: Fundamentos de Clase e Obxectos

- Define unha clase chamada "Rectangulo" cun construtor que reciba os parámetros: "lonxitude" e "ancho". Gárdaos en atributos de instancia.
- Engade un método chamado "area" que calcule e devolva a área do rectángulo.
- Crea dous obxectos (instancias) de "Rectangulo" con diferentes medidas, calcula a súa área e imprime o seu diccionario interno de atributos.


```python
#nome clase 
class Rectangulo():
    def __init__(self, longitud, ancho): #contructor
        self.longitud = longitud
        self.ancho =ancho

#metodo
    def area(self):
        return self.longitud * self.ancho
    

#objetos
rect1= Rectangulo(5,9)
rect2 = Rectangulo(7,3)

#calcular areas
print("área rectangulo", rect1.area)
print("área rectangulo", rect2.area)

#imprimir diccionario interno
print("atributos:" ,rect1.__dict__)
print("atributos:" ,rect2.__dict__)
```

### Exercicio 2: Atributos "Privados" e Propiedades

- Crea unha clase "ContaBancaria".
- No constructor (`__init__`), inicializa un atributo "privado" denominado "saldo". Que non se poida acceder directamente a el.
- Crea un "getter" para poder consultar dito saldo.
- Crea un "setter" que, ao intentar modificar o saldo, só permita asignar valores positivos e "racionales".
- Fai probas de uso.

```python
#nome clase

class ContaBancaria():
    def __init__(self,saldo_inicial=0):
        self.__saldo = 0
        
    self.saldo= saldo_inicial

    #getter
    @property
    def saldo(self):
        return self.__saldo
    
    #setter 
    @saldo.setter
    def saldo(self, nuevo_saldo):
        if isinstance(nuevo_saldo, (int, float)) and nuevo_saldo >=0
            self.__saldo=nuevo_saldo
        else:
            print("el saldo debe ser un numero postivo")    

#pruebas

 conta1 = ContaBancaria(100)
conta1.saldo = 200
print(conta1.saldo)  # Correcto → 200

conta1.saldo = -50   # Incorrecto
conta1.saldo = "ola" # Incorrecto

print(conta1.__saldo) 


```

### Exercicio 5: Métodos Estáticos, de Clase e de Instancia

- Crea unha clase denominada "UtilidadesMatematicas" cos seguintes métodos:
    - Un método estático denominado "e_par" que devolva True ou False.
    - Un método de clase denominado "informacion" que devolva unha cadea descritiva sobre a clase.
    - Un método de instancia denominado "potencia" que, para un número base almacenado nun atributo de instancia inicializado no constructor, calcule a potencia a dito exponente.
- Fai o preciso para chamar a todos os métodos creados.

```python
#nombre clase
class  UtilidadesMatematicas: 

    def __init__(self,base):
         self.base = base 
    
    #metodo estatico
    @staticmethod
    def e_par(numero):
        return numero % 2== 0
    
    #metodo de clase
    @classmethod
    def informacion(cls):
        return f"la clase  {cls.__name__} son métodos matemáticos" 
    
    #metodo de instancia
    def potencia(self, exponente):
        return self.base** exponente


#llamada a los metodos
print (UtilidadesMatematicas.e_par(1)) # false
print (UtilidadesMatematicas.e_par(5)) # true


print(UtilidadesMatematicas.informacion())

u = UtilidadesMatematicas(3)

print(u.potencia(4))  #3*3*3*3=81
```

### Exercicio 6: Dunder Methods

- Crea unha clase "Vector2D" para representar un vector no plano cun constructor que reciba as coordenadas "x" e "y".
- Implementa (ou sobrescribe) o método "__add__" para que poidas sumar dous vectores usando o operador "+". Deberá devolver un obxecto novo coa suma das coordenadas.
- Implementa (ou sobrescribe) o método "__str__" para que, cando se convirta o vector a string, amose un formato lexible como "Vector2D(x=5, y=3)".
- Implementa (ou sobrescribe) o método "__eq__" para que dous vectores sexan iguais se as súas coordenadas "x" e "y" coinciden.
- Constrúe exemplos de uso.

```python
#nombre clase

class Vector2D:
    def __init__(self, x, y):
        self.x= x
        self.y =y 

    #metodo add
    def __add__(self, other):
        return Vector2D(self.x+ other.x, self.y+ other.y)
    
    #metodo str
    def __str__(self):
        return f"Vector2D(x={self.x}, y={self.y})"
    
    #metodob eq

    def __eq__(self, other):
        return self.x ==other.x and self.y == other.y



# Exemplos de uso
v1 = Vector2D(2, 3)
v2 = Vector2D(4, 1)

# Usar o operador +
v3 = v1 + v2
print(v3)  # Vector2D(x=6, y=4)

# Mostrar un vector
print(v1)  # Vector2D(x=2, y=3)

# Comparar vectores
v4 = Vector2D(2, 3)
print(v1 == v4)  # True
print(v1 == v2)  # False
            




```

```python


#nombre clase 

class Imprimible: 

    def amosar(self):
        return "Son un obxeto imprimible"



#partiendo de esta clase

class Pila():
    # Constructor
    def __init__(self):
        self.__stack_list = [] # Atributo "privado"

    def push(self, val):
        self.__stack_list.append(val) # Engadimos un valor a este atributo privado

    def pop(self):
        if self.__stack_list == []:
            raise IndexError("Pop en lista baleira")
        else:
            val = self.__stack_list[-1]
            del self.__stack_list[-1]
            return val

mistack = Pila()
mistack._Pila__stack_list  # Así SI se pode acceder a __stack_list, aínda que tampouco tería moito sentido facelo
mistack.__stack_list # Así NON, da un erro: AttributeError: 'Pila' object has no attribute '__stack_list'


# y de esta

class PilaSuma(Pila):
    def __init__(self): # Constructor
        super().__init__()  # super permite acceder aos métodos da/s clase/s da que herda
        self.__suma = 0 # Definimos un atributo denominado suma

    @property # Definimos unha propiedade para poder acceder ao atributo directamente
    def suma(self):
        return self.__suma

    def push(self, val): # Estamos como redefinindo o método push da clase pai
        self.__suma += val
        super().push(val)

    def pop(self): # E facemos o propio con pop
        val = super().pop()
        self.__suma -= val
        return val

mipsuma = PilaSuma()
mipsuma.suma
    
#comprobar relación 

print(issubclass(PilaSegura, Pila))        # True
print(isinstance(pila_segura, PilaSegura)) # True
print(isinstance(pila_segura, Pila))       # True


#nueva clase 

class PilaSumaImprimible(PilaSuma,imprimible):

   def __init__(self):
        super().__init__()

    # Sobrescribimos amosar
    def amosar(self):
        return f"PilaSuma con suma total: {self.suma}"

p = PilaSumaImprimible()

p.push(4)
p.push(6)
p.push(2)

print(p.amosar())  # PilaSuma coa suma total: 12

print(p.pop())     # elimina 2
print(p.amosar())  # suma agora 10


# comprobar relación
print(issubclass(PilaSumaImprimible, PilaSuma))   # True
print(issubclass(PilaSumaImprimible, Imprimible)) # True

print(isinstance(p, PilaSumaImprimible)) # True
print(isinstance(p, Pila)) # True


# dunder methods
print(PilaSumaImprimible.__mro__)  #deternina el orden en que se buscan los emtodos y los atributos
print(PilaSumaImprimible.__bases__)  #permite alamcenar ,consultar, atributos o metodos en jerarquia de clases
```
