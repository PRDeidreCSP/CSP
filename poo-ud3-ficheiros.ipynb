### Apertura Segura con try...except:

* Escribe unha función chamada `abrir_ficheiro_seguro(nome_ficheiro, modo)` que intente abrir un ficheiro no modo especificado.
* Debe usar un bloque `try...except` para capturar calquera excepción (como `FileNotFoundError`).
* Se a apertura é exitosa, a función debe devolver o obxecto do ficheiro. Se hai un erro, debe amosar a mensaxe de erro e devolver `None`.
* Fai probas invocando a función cun ficheiro que non exista e despois cun que si exista (por exemplo, en modo lectura `"r"`). Non esquezas pechar o ficheiro se se abriu correctamente.

````python
def abrir_ficheiro_seguro(nome_ficheiro, modo):

    try :
        fichero = open(nome_ficheiro, modo)
        print ("fichero abierto")
        return fichero
    except FileNotFoundError as error:
        print(f"error: el fichero no existe {error}")
        return None


# PRUEBAS

#FICHERO QUE NO EXISTE 

f1= abrir_ficheiro_seguro("fichero.txt", "r")
print(f1)

#fichero existente (lo he credo previamente)
f2 = abrir_ficheiro_seguro("fichero_pruebas.txt","r")

for linea in f2:
    print(linea, end="")
f2.close()
````

### Analizador de Ficheiro de Texto

* Pídelle ao usuario o nome dun ficheiro de texto.
* O programa debe abrir o ficheiro e calcular:
    * Número total de liñas.
    * Número total de palabras.
    * Número total de caracteres.
* Crea un ficheiro `proba.txt` con algunhas liñas de texto e comproba que os cálculos son correctos.

````python
def analizar_fichero(nombre_fichero):
    try:
        with open(nombre_fichero, "r") as f:
            lineas = f.readlines()

            num_lineas = len(lineas)
            num_palabras = 0
            num_caracteres = 0

            for linea in lineas:
                num_palabras += len(linea.split())
                num_caracteres += len(linea)

        print("\n--- RESULTADOS ---")
        print("Líneas:", num_lineas)
        print("Palabras:", num_palabras)
        print("Caracteres:", num_caracteres)

    except FileNotFoundError:
        print("Error: el fichero no existe.")


# --- Crear fichero de prueba ---
with open("pruebas.txt", "w") as f:
    f.write("Hola mundo\n")
    f.write("Este es un fichero de prueba\n")
    f.write("Python es divertido\n")


# --- Pedir nombre al usuario ---
nombre = input("Introduce el nombre del fichero: ")

# --- Analizar el fichero ---
analizar_fichero(nombre)

````

### Copiador de Ficheiros con Número de Liña

* Pídelle ao usuario o nome dun ficheiro de orixe.
* Crea un novo ficheiro denominado `copia_con_numeros.txt`.
* Le o ficheiro de orixe **liña por liña** e escribe cada liña no novo ficheiro, pero **engadíndolle o número de liña ao principio**.
* Por exemplo, se unha liña orixinal é "Hola mundo", no novo ficheiro debe aparecer "1 Hola mundo".

````python
def copiar_con_numeros():
    nombre = input("Introduce el nombre del fichero de origen: ")

    try:
        with open(nombre, "r") as origen, open("copia_con_numeros.txt", "w") as copia:
            num_linea = 1

            for linea in origen:
                copia.write(f"{num_linea} {linea}")
                num_linea += 1

        print("Copia realizada correctamente en 'copia_con_numeros.txt'.")

    except FileNotFoundError:
        print("Error: el fichero de origen no existe.")


# Ejecutar programa
copiar_con_numeros()
````

###  Xestor de Contactos en JSON

* Crea un programa para xestionar unha axenda de contactos nun ficheiro de texto `contactos.JSON`.
* A estrutura será unha lista de dicionarios, onde cada dicionario ten as chaves `nome`, `telefono`, `email`.
* O programa debe ter un menú para:
    * **Engadir contacto**: Pide os datos e engádeos ao ficheiro.
    * **Buscar contacto**: Pide un nome e amosa o teléfono e email de todas as liñas que conteñan ese nome.
    * **Listar todos**: Amosa todos os contactos numerados.

````python
import json
import os

archivo = "contactos.json"

# --- Cargar contactos existentes ---
if os.path.exists(archivo):
    with open(archivo, "r") as f:
        try:
            contactos = json.load(f)
        except json.JSONDecodeError:
            contactos = []
else:
    contactos = []

# --- Funciones ---
def guardar_contactos():
    with open(archivo, "w") as f:
        json.dump(contactos, f, indent=4)

def añadir_contacto():
    nombre = input("Nombre: ")
    telefono = input("Teléfono: ")
    email = input("Email: ")
    contactos.append({"nome": nombre, "telefono": telefono, "email": email})
    guardar_contactos()
    print("Contacto añadido.\n")

def buscar_contacto():
    busqueda = input("Introduce el nombre a buscar: ").lower()
    encontrados = [c for c in contactos if busqueda in c["nome"].lower()]
    if encontrados:
        for c in encontrados:
            print(f"Nombre: {c['nome']}, Teléfono: {c['telefono']}, Email: {c['email']}")
    else:
        print("No se encontró ningún contacto.\n")

def listar_todos():
    if contactos:
        for i, c in enumerate(contactos, start=1):
            print(f"{i}. Nombre: {c['nome']}, Teléfono: {c['telefono']}, Email: {c['email']}")
    else:
        print("No hay contactos en la agenda.\n")

# --- Menú principal ---
while True:
    print("\n--- MENÚ CONTACTOS ---")
    print("1. Añadir contacto")
    print("2. Buscar contacto")
    print("3. Listar todos")
    print("4. Salir")
    opcion = input("Elige una opción: ")

    if opcion == "1":
        añadir_contacto()
    elif opcion == "2":
        buscar_contacto()
    elif opcion == "3":
        listar_todos()
    elif opcion == "4":
        print("Saliendo...")
        break
    else:
        print("Opción no válida. Intenta de nuevo.")



````

### Cálculo con Datos Numéricos usando NumPy

* Crea un ficheiro de texto chamado `matriz.txt` con varias liñas de números separados por espazos. Por exemplo:
    ```
    1.5 2.3 4.8
    5.1 3.2 7.4
    9.6 8.7 6.5
    ```
* Escribe un script que:
    1. Cargue os datos do ficheiro nun array de **NumPy**.
    2. Calcule e amose por pantalla: a media de todos os números, a suma de cada fila e o valor máximo de cada columna. 
    3. Finalmente, garde o array orixinal multiplicado por 2 nun novo ficheiro `matriz_dobre.txt` con formato `'%.2f'`.

````python
import numpy as np

# --- Crear fichero de prueba ---
with open("matriz.txt", "w") as f:
    f.write("1.5 2.3 4.8\n")
    f.write("5.1 3.2 7.4\n")
    f.write("9.6 8.7 6.5\n")

# --- Cargar datos con NumPy ---
try:
    matriz = np.loadtxt("matriz.txt")
except Exception as e:
    print("Error al leer el fichero:", e)
    exit()

# --- Cálculos ---
media_total = np.mean(matriz)
suma_filas = np.sum(matriz, axis=1)
max_columnas = np.max(matriz, axis=0)

print("--- RESULTADOS ---")
print("Media de todos los números:", media_total)
print("Suma de cada fila:", suma_filas)
print("Valor máximo de cada columna:", max_columnas)

# --- Guardar matriz multiplicada por 2 ---
matriz_doble = matriz * 2
np.savetxt("matriz_dobre.txt", matriz_doble, fmt="%.2f")

print("Matriz multiplicada por 2 guardada en 'matriz_dobre.txt'.")
````
